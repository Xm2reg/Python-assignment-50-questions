```python
n = int(input("Enter the number of values will be provided:"))
li = [input(f"Enter your input {i+1}:") for i in range(n)]
final = []
for x in li:
    if x not in final:
        final.append(x)
    else:
        continue
print(f"Original List provided : {li}")
print(f"final list with removing dulicates: {final}")
```
