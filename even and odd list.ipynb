```python
li = [int(x) for x in input("Enter the values separated by comma:").split(',')]

new_li = sorted(set(li))
even = []
odd = []
for x in new_li:
    if x % 2 == 0:
        even.append(x)
    else:
        odd.append(x)

print(f"List of even numbers:{even}")
print(f"List of odd numbers:{odd}")
```
