```python
while True:
    try:
        nums = [int(x) for x in input(f"Enter numbers separated by comma: ").split(',')]
        unique_nums = sorted(set(nums), reverse=True)
        if len(unique_nums) >= 2:
            print(f"Second largest number: {unique_nums[1]}")
        else:
            print(f"\nAll numbers are identical. The largest is: {unique_nums[0]}")
            
        break 
        
    except ValueError:
        print("Invalid input! Please enter numbers only.\n")
```
