```python
li = [int(x) for x in input(f"Enter {n} values separated by spaces: ").split()]

unique_sorted_list = sorted(set(li))

print(f"\nValues provided: {li}")

if len(unique_sorted_list) == 2:
    print(f"Second largest number is: {unique_sorted_list[1]}")
    print(f"Second smallest number is: {unique_sorted_list[0]}")
elif len(unique_sorted_list) > 2 :
    print(f"Second largest number is: {unique_sorted_list[-2]}")
    print(f"Second smallest number is: {unique_sorted_list[1]}")
else:
    print("Not enough numbers to compare")
```
