```python
marks = [float(input(f"Enter subject {i+1} marks: ")) for i in range(5)]
total_marks = sum(marks)
pr = round((total_marks / 500) * 100,2)

print(f"Total obtained marks: {total_marks}")
print(f"Percentage obtained: {pr}%")

if pr >= 90:
    grade = "O"
elif pr >= 80:
    grade = "A"
elif pr >= 70:
    grade = "B"
elif pr >= 60:
    grade = "C"
elif pr >= 50:
    grade = "D"
else:
    grade = "F"

print(f"Grade is: {grade}")
```
