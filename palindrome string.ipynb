```python
while True:
    try:
        sample = input("Enter your string:")
        if sample.isdigit():
            print("Enter valid String")
            continue
        break
          
    except ValueError:
        print("Invalid input!, enter valid string.")

rev = sample[::-1]

if sample == rev:
    print("Provided string is Palindrome.")
else:
    print("Provided string is not Palindrome.")


    
```
