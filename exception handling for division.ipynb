```python
while True:
    try:
        n1 = int(input("enter numerator:"))
        n2 = int(input("enter Denomenator:"))
        if n2 == 0:
            print("Denominator cannot be zero.")
        else:
            result = n1/n2
            break
    except ValueError:
        print("Invalid input!. Enter numerical value")
print(f"{n1} divided by {n2} is {result}")
```
