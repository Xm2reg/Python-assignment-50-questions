```python
while True:
    try:
        unit = float(input("Enter number of units consumed: "))
        if unit < 0:
            print("Units cannot be negative. Please enter a valid number.")
            continue 
        break
    except ValueError:
        print("Invalid Input! Enter a valid numerical value.")


if unit <= 100:
    bill = unit * 4.00
elif unit <= 200:
    bill = (100 * 4.00) + ((unit - 100) * 5.50)
elif unit <= 300:
    bill = (100 * 4.00) + (100 * 5.50) + ((unit - 200) * 6.50)
else:    
    bill = (100 * 4.00) + (100 * 5.50) + (100 * 6.50) + ((unit - 300) * 7.50)

print(f"Electricity bill is: {bill:.2f} rupees")
```
