```python
while True:
    try:
        nums = [float(input(f"Enter number {i+1}: ")) for i in range(3)]
        unique_nums = sorted(set(nums), reverse=True)
        if len(unique_nums) >= 2:
            print(f"\nLargest number: {unique_nums[0]}")
            print(f"Second largest number: {unique_nums[1]}")
        else:
            print(f"\nAll three numbers are identical. The largest is: {unique_nums[0]}")
            
        break 
        
    except ValueError:
        print("Invalid input! Please enter numbers only.\n")
```
