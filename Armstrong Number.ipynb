```python
while True:
    try:
        num = int(input("Enter a positive whole number: "))
        if num < 0:
            print("Please enter a positive value.")
            continue
        break
    except ValueError:
        print("Invalid input! Please enter a whole number.")
        
num_str = str(num)
power = len(num_str)

armstrong_sum = sum(int(digit) ** power for digit in num_str)

if armstrong_sum == num:
    print(f"\n{num} is an Armstrong number!")
else:
    print(f"\n{num} is not an Armstrong number.")
```
