```python
while True:
    us = input("Enter your string: ")    
    if not us.replace(" ", "").isalpha():
        print("Please enter a valid string, no numbers or symbols allowed.")
        continue
    break
    
us = us.replace(" ", "")

print(f"\nYour string without spaces: '{us}'")
print(f"Total characters: {len(us)}")


```
