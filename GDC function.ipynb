```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

limit = min(num1, num2)

common_factors = [i for i in range(1, limit + 1) if num1 % i == 0 and num2 % i == 0]

gcd = max(common_factors)

print(f"The common factors are: {common_factors}")
print(f"The GCD of {num1} and {num2} is: {gcd}")
```
