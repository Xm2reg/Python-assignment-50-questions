```python
def uni_merge():
    list1 = input("Enter first list of elements separated by comma: ").split(',')
    list2 = input("Enter second list of elements separated by comma: ").split(',')
    
    merged_list = list1 + list2
    
    final_list = list(dict.fromkeys(merged_list))
    
    print(f"Final list: {final_list}")
    

    
            

```
