```python
try:
    li = input("Enter your sentence:").split()
    print(f"Maximum index search avaliable in provided sentence is {len(li)-1}")
    
    n = int(input("Enter the index for value fetching:"))
    val = li[n]
    
    print(f"Value present at index {n} is {val}")
except IndexError:
    print("Invalid Input!. plz enter index value as per limit provided.")
```
